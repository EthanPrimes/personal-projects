```python
import pyperclip

# Reverses each word in a snippet
PUNCTUATION = [".", ",", "?", "!"]
string = "?uoy sey ,ti dnatsrednu I .ylralimis dnopser lliw I ."
string = string.lower()

def reverse_word(full_word):
    final_char = full_word[-1]
    if final_char in PUNCTUATION:
        return full_word[:-1][::-1] + final_char
    return full_word[::-1]

res = " ".join(reverse_word(word) for word in string.split())
pyperclip.copy(res)
print(res)
```
