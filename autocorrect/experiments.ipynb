# Levenshtein Distance

```python
from collections import defaultdict

import numpy as np
```

```python
def levenshtein_distance(s, t):
    """
    Computes the Levenshtein distance between s and t.

    Uses the DP approach from Wikipedia's pseudocode.
    """
    m, n = len(s), len(t)
    d = np.zeros((m+1, n+1))
    d[:, 0] = np.array([i for i in range(m+1)])
    d[0, :] = np.array([i for i in range(n+1)])
    for i in range(1, m+1):
        for j in range(1, n+1):
            if s[i-1] == t[j-1]:
                substitution_cost = 0
            else:
                substitution_cost = 1

            d[i, j] = min(
                d[i-1, j] + 1,
                d[i, j-1] + 1,
                d[i-1, j-1] + substitution_cost
            )

    return d[m, n]
```

```python
assert levenshtein_distance("kitten", "sitting") == 3
assert levenshtein_distance("testing", "") == 7
assert levenshtein_distance("rad", "sky") == 3
assert levenshtein_distance("", "a prime determinant") == 19
```

# Reading in the 100k most common English words

```python
WORDS_FILE = "google_books-100k.txt"
with open("data/" + WORDS_FILE, "r") as file:
    all_words = [line.strip().lower() for line in file if "#" not in line]
```

# Running the Autocorrect

```python
misspelled_word = "warpped"
nearby_words = sorted(
    [(s, levenshtein_distance(misspelled_word, s)) for s in all_words],
    key=lambda x: x[1],
)

# Get all the words that are minimal distance from the misspelled word
min_distance = nearby_words[0][1]
closest_words = []
i = 0
while (nearby_words[i][1] == min_distance) or i < 10:
    closest_words.append(nearby_words[i])
    i += 1

print(closest_words)
```
