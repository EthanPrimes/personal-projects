# Register Machines
Register machines are essentially tapes storing nonnegative integers. We will create a class that acts as a single register and will define functions that can operate on them and combine them in various ways.

```python
class Register():
    def __init__(self, register=None):
        if not register:
            register = []
        self.register = register
```
